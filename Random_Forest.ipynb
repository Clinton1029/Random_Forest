## 1. Importation of Libraries

```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.utils import resample
from sklearn.ensemble import RandomForestClassifier

```

## 2. Loading The Data

```python
# Read the Excel file
df = pd.read_excel("/content/Bank_Loan_Dataset.xlsx")

# Display the dataset
df
```

## 3. Encode the Categorical Variables

```python
encoder = LabelEncoder()

df["House"] = encoder.fit_transform(df["House"])
df["Loan"] = encoder.fit_transform(df["Loan"])

print(df)
```

## 4. Select Features and Target

```python
X = df[["Age", "Income", "House"]]

y = df["Loan"]
```

## 5. Perform Bootstrap Sampling

```python
bootstrap_samples = []

for i in range(3):

    sample = resample(
        df,
        replace=True,
        n_samples=5,
        random_state=i
    )

    bootstrap_samples.append(sample)

    print(f"\nBootstrap Sample {i+1}")

    print(sample)
```

## 6. Build the Random Forest

```python
model = RandomForestClassifier(

    n_estimators=3,
    bootstrap=True,
    max_depth=2,
    max_samples=5,
    random_state=1

)

```

## 7. Train Model

```python
model.fit(X,y)
```

## 8. Display the Prediction

```python
new_customer = pd.DataFrame({
    "Age": [38],
    "Income": [75],
    "House": [1]
})

prediction = model.predict(new_customer)

if prediction == 1:
    print("Prediction: Loan Approved")
else:
    print("Prediction: Loan Not Approved")
```
